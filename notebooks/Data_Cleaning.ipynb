## Datatype Conversion
Converting date columns from object datatype to datetime format for time-based analysis and feature engineering.

## Null Value Handling

Handling missing values in the products dataset to maintain data consistency and avoid null values during analysis and dashboard creation.

```python
# Fill missing product category names

products['product_category_name'] = (
    products['product_category_name']
    .fillna('Unknown')
)

# Fill numeric null values with median

products['product_name_lenght'] = (
    products['product_name_lenght']
    .fillna(products['product_name_lenght'].median())
)

products['product_description_lenght'] = (
    products['product_description_lenght']
    .fillna(products['product_description_lenght'].median())
)

products['product_photos_qty'] = (
    products['product_photos_qty']
    .fillna(products['product_photos_qty'].median())
)

products['product_weight_g'] = (
    products['product_weight_g']
    .fillna(products['product_weight_g'].median())
)

products['product_length_cm'] = (
    products['product_length_cm']
    .fillna(products['product_length_cm'].median())
)

products['product_height_cm'] = (
    products['product_height_cm']
    .fillna(products['product_height_cm'].median())
)

products['product_width_cm'] = (
    products['product_width_cm']
    .fillna(products['product_width_cm'].median())
)

# Verify null values
products.isnull().sum()
```

## Feature Engineering

Creating business features for sales, delivery, and operational analysis.

### Delivery Delay
Calculating the difference between actual delivery date and estimated delivery date to measure late or early delivery.

```python
# Delivery delay in days
orders['delivery_delay'] = (
    orders['order_delivered_customer_date'] - orders['order_estimated_delivery_date']
).dt.days
```

### Lead Time
Calculating the number of days taken from purchase to final delivery.

```python
# Lead time in days
orders['lead_time'] = (
    orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']
).dt.days
```

### Total Sales
Creating total sales per order item using product price and freight value.

```python
# Total sales per order item
order_items['total_sales'] = order_items['price'] + order_items['freight_value']
```

### Purchase Year
Extracting the year from the order purchase timestamp.

```python
# Purchase year
orders['purchase_year'] = orders['order_purchase_timestamp'].dt.year
```

### Purchase Month
Extracting the month from the order purchase timestamp.

```python
# Purchase month
orders['purchase_month'] = orders['order_purchase_timestamp'].dt.month
```

### Purchase Day
Extracting the day from the order purchase timestamp.

```python
# Purchase day
orders['purchase_day'] = orders['order_purchase_timestamp'].dt.day
```

### Purchase Hour
Extracting the hour from the order purchase timestamp to analyze peak ordering time.

```python
# Purchase hour
orders['purchase_hour'] = orders['order_purchase_timestamp'].dt.hour
```

# Push Cleaned Data into MySQL
Loading the cleaned Olist datasets into MySQL as separate staging tables for warehouse modeling, SQL analysis, and downstream ML and Power BI use.

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from urllib.parse import quote_plus


# MySQL database connection configuration

load_dotenv(dotenv_path="D:\Resume Projects\supply-chain-analytics-pipeline\.env")  #load .env file

user     = os.getenv("DB_USER")
password = quote_plus(os.getenv("DB_PASSWORD"))
host     = os.getenv("DB_HOST")
port     = os.getenv("DB_PORT")
database = os.getenv("DB_NAME")

# Create engine
engine = create_engine(f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}")
```

```python
# Push cleaned tables separately to MySQL

orders.to_sql('orders', con=engine, if_exists='replace', index=False)
order_items.to_sql('order_items', con=engine, if_exists='replace', index=False)
order_payments.to_sql('order_payments', con=engine, if_exists='replace', index=False)
customers.to_sql('customers', con=engine, if_exists='replace', index=False)
products.to_sql('products', con=engine, if_exists='replace', index=False)
sellers.to_sql('sellers', con=engine, if_exists='replace', index=False)
geolocation.to_sql('geolocation', con=engine, if_exists='replace', index=False)
order_reviews.to_sql('order_reviews', con=engine, if_exists='replace', index=False)

print("All cleaned tables pushed to MySQL successfully.")
```
